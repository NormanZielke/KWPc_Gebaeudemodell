```python
from pathlib import Path
import geopandas as gpd
import pandas as pd
import matplotlib.pyplot as plt
import plotly.graph_objects as go
```

```python
GPKG_PATH = (
    "1_Rohdaten/HN/HN-Gebäudemodell/"
    "HN-Gebäudemodell_04_08_2026/"
    "260728_Gebäudemodell_HohenNeuendorf.gpkg"
)
```

```python
gdf = gpd.read_file(GPKG_PATH)
gdf
```

```python
demand_kwh_ges = gdf["demand_kwh"].sum()*1e-6
print("Wärmeverbrauch über alle Einträge beträgt", round(demand_kwh_ges,2), "GWh")
```

```python
gdf_with_ID = gdf[~gdf["GebaeudeID"].isna()]
demand_kwh_ID = gdf_with_ID["demand_kwh"].sum()*1e-6
print("Wärmeverbrauch über alle Gebäude_ID beträgt", round(demand_kwh_ID,2), "GWh")
```

```python
gdf_with_funktion = gdf[~gdf["funktion"].isna()]
df = gdf[~gdf["funktion"].isna() & ~gdf["GebaeudeID"].isna()]
print(df)
print("Es gibt keine Gebäude, die ein Funktion und eine Gebäude_ID haben")
```
