## Gebäudemodell HN 

```python
GPKG_PATH = (
    "1_Rohdaten/HN/HN-Gebäudemodell/"
    "HN-Gebäudemodell_04_08_2026/"
    "260728_Gebäudemodell_HohenNeuendorf.gpkg"
)
```

```python
gdf = gpd.read_file(GPKG_PATH)
```

```python
demand_kwh_ges = gdf["demand_kwh"].sum()*1e-6
print("Wärmeverbrauch über alle Einträge beträgt", round(demand_kwh_ges,2), "GWh")
```

```python
gdf_with_ID = gdf[~gdf["GebaeudeID"].isna()]
demand_kwh_ID = gdf_with_ID["demand_kwh"].sum()*1e-6
gdf_with_function = gdf[~gdf["funktion"].isna()]
demand_kwh_function = gdf_with_function["demand_kwh"].sum()*1e-6
print("Wärmeverbrauch über alle Gebäude_ID beträgt", round(demand_kwh_ID,2), "GWh")
print("Es gibt",len(gdf_with_ID), "mit Gebaeude_ID")
print("Wärmeverbrauch über alle Gebäude ohne Gebäude_ID und mit funktion beträgt", round(round(demand_kwh_ges,2) - round(demand_kwh_ID,2),2), "GWh")
print("Wärmeverbrauch über alle Gebäude ohne Gebäude_ID und mit funktion beträgt", round(demand_kwh_function,2), "GWh")
print("Es gibt",len(gdf_with_function), "ohne Gebaeude_ID")
```
